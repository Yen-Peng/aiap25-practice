# EDA

This notebook will be used for exploratory data analysis for the hotel booking cancellation risk prediction project.

### Section 1: Import dependencies:

```python
import sqlite3
import pandas as pd
import numpy as np
import seaborn as sb
import matplotlib.pyplot as pp
```

### Section 2: Analyze data in hotel records

##### 2.1. Analyze bookings table

```python
## connect to db
conn = sqlite3.connect('hotel.db')
cursor = conn.cursor()

## check tables in db
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
print("Tables in db", cursor.fetchall())
```

```python
# preview table data structure
df_bookings = pd.read_sql_query("SELECT * FROM bookings", conn)
df_bookings.head()
```

```python
# preview table summary info => all non-null values, skip missing values check
df_bookings.info()
```

```python
# check descriptive statistics
df_bookings.describe().T
```

```python
# check duplicate record count => 353 duplicate records
df_bookings.duplicated().sum()
```

```python
# sort and display duplicated records
df_bookings[df_bookings.duplicated(keep=False)].sort_values(by=list(df_bookings.columns))
```

```python
# check consistency of is_canceled column => inconsistent values found, should be 0 or 1
df_bookings['is_canceled'].unique()
```

##### 2.2. Analyze reviews table

```python
# preview table's data structure
df_reviews = pd.read_sql_query("SELECT * FROM reviews", conn)
df_reviews.head()
```

```python
# preview table summary info => all non-null values, skip missing values check
df_reviews.info()
```

```python
# check descriptive statistics
df_reviews.describe().T
```

```python
# check duplicate record count => 51 duplicate records
df_reviews.duplicated().sum()
```

```python
df_reviews[df_reviews.duplicated(keep=False)].sort_values(by=list(df_reviews.columns))
```

```python
# remove duplicate records from both tables
df_bookings.drop_duplicates(inplace=True)
df_reviews.drop_duplicates(inplace=True)
```
