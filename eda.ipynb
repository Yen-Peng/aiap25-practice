# EDA

This notebook will be used for exploratory data analysis for the hotel booking cancellation risk prediction project.

Import dependencies:

```python
import sqlite3
import pandas as pd
import numpy as np
import seaborn as sb
import matplotlib.pyplot as pp
```

Check tables in hotel database records:

```python
conn = sqlite3.connect('hotel.db')
cursor = conn.cursor()

cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
print("Tables in db", cursor.fetchall())
```

Preview bookings table's data structure and summary:

```python
df_bookings = pd.read_sql_query("SELECT * FROM bookings", conn)
df_bookings.head()
```

```python
df_bookings.info()
```

Preview reviews table's data structure and summary:

```python
df_reviews = pd.read_sql_query("SELECT * FROM reviews", conn)
df_reviews.head()
```

```python
df_reviews.info()
```

Detect missing values:

```python
df_bookings.isnull().sum()
```

```python
df_reviews.isnull().sum()
```

Check descriptive statistics:

```python
df_bookings.describe().T
```

```python
df_reviews.describe().T
```
